```python
# Heading 1 (Broad context / Top-level block / Script section)
# Deterministic ontology intersection validation and profiling module.

import os
from typing import Dict, Any, List, Set, Tuple
import pandas as pd
import pronto

from OmniPhysiBoSS.utils.logger import get_custom_logger

logger = get_custom_logger(__name__)

def filter_markers_by_go_validity(
    marker_df: pd.DataFrame, 
    go_ontology_path: str
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Filter a cell marker dataframe to retain exclusively records containing a cellontology_id 
    or cell_name cross-referenced within the Gene Ontology, returning the clean dataset and an inspection profile.

    :param marker_df: Pandas DataFrame containing raw CellMarker 2.0 database entries.
    :type marker_df: pd.DataFrame
    :param go_ontology_path: Filesystem path targeting the comprehensive go-basic.obo file asset.
    :type go_ontology_path: str
    :return: A tuple containing the filtered markers DataFrame and a descriptive GO terms context profile DataFrame.
    :rtype: Tuple[pd.DataFrame, pd.DataFrame]
    :raises FileNotFoundError: If the target OBO file trajectory cannot be resolved on the filesystem.
    """
    # Heading 1 (Broad context / Top-level block / Script section)
    # File handling, validation, and raw pronto parsing operations
    logger.info("Initializing synchronous Gene Ontology scanning and filtering execution loop.")

    if not os.path.exists(go_ontology_path):
        logger.error("Target ontology lookup path is unreachable: %s", go_ontology_path, exc_info=True)
        raise FileNotFoundError(f"Ontology file missing at target path: {go_ontology_path}")

    logger.debug("Loading complete OBO graph structure via pronto from tracking path: %s", go_ontology_path)
    cl_ontology = pronto.Ontology(go_ontology_path)
    
    # Standardize input dataframe formats upfront to guarantee join matching
    working_df = marker_df.copy()
    working_df['cellontology_id'] = working_df['cellontology_id'].astype(str).str.replace("_", ":").str.strip()
    
    ## Heading 2 (Specific operation / Sub-step inside a loop or function / Secondary logic)
    ## Extract and pre-index unique cell configurations to optimize search steps
    cell_reference = working_df[['cellontology_id', 'cell_name']].drop_duplicates()
    
    # Pre-compute query profiles for each unique cell line to avoid repetitive string manipulation inside the loop
    cell_queries: Dict[str, Dict[str, str]] = {}
    for _, row in cell_reference.iterrows():
        cl_id = str(row['cellontology_id'])
        raw_name = str(row['cell_name'])
        
        norm = raw_name.lower().strip()
        norm_stripped = norm.replace("cells", "").replace("cell", "").strip()
        fallback_query = "natural killer" if norm_stripped == "nk" else norm_stripped
        
        cell_queries[cl_id] = {
            "raw_name": raw_name,
            "lower_name": norm,
            "fallback_query": fallback_query
        }

    go_profile_records: List[Dict[str, str]] = []
    mapped_cl_ids: Set[str] = set()

    ## Heading 2 (Specific operation / Sub-step inside a loop or function / Secondary logic)
    ## Single-pass parse of the entire ontology structure to isolate functional namespaces and metrics
    for term in cl_ontology.terms():
        if not term.id.startswith("GO:"):
            continue

        term_name = str(term.name) if term.name else ""
        term_name_lower = term_name.lower()
        term_name_norm = term_name_lower.replace("cell", "").strip()
        
        term_namespace = str(term.namespace) if term.namespace else "Unknown Namespace"
        term_definition = str(term.definition) if term.definition else "No definition block available."
        
        # Isolate semantic explicit CL tokens if present (fallback for go-plus environments)
        term_cl_tokens: Set[str] = set()
        if term.xrefs:
            for xref in term.xrefs:
                xref_norm = xref.id.replace("_", ":").upper().strip()
                if "CL:" in xref_norm:
                    term_cl_tokens.add(xref_norm[xref_norm.find("CL:"):].split()[0])
        if term.relationships:
            for rel, related_terms in term.relationships.items():
                for rel_term in related_terms:
                    rel_norm = rel_term.id.replace("_", ":").upper().strip()
                    if "CL:" in rel_norm:
                        term_cl_tokens.add(rel_norm[rel_norm.find("CL:"):])

        ### Heading 3 (Deep sub-step / Conditional branch / Granular execution detail)
        ### Evaluate text inclusion matrices against the active GO term properties
        for cl_id, info in cell_queries.items():
            is_match = False
            
            # Scenario A: Explicit cross-reference token matching
            if cl_id in term_cl_tokens:
                is_match = True
                
            # Scenario B: Direct inclusion of full cell name string (e.g., "t cell" inside "t cell activation")
            if not is_match and info["lower_name"] in term_name_lower:
                is_match = True
                
            # Scenario C: Textual fallback canonical parsing matching definitions
            if not is_match and term_name_norm:
                q = info["fallback_query"]
                if (q == term_name_norm) or (len(q) > 3 and q in term_name_norm):
                    is_match = True

            ### Heading 3 (Deep sub-step / Conditional branch / Granular execution detail)
            ### Log found alignments and compile records to final mapping dictionaries
            if is_match:
                mapped_cl_ids.add(cl_id)
                go_profile_records.append({
                    "cellontology_id": cl_id,
                    "cell_type_name": info["raw_name"],
                    "go_term_id": str(term.id),
                    "go_term_name": term_name,
                    "go_namespace": term_namespace,
                    "go_definition": term_definition
                })

    ## Heading 2 (Specific operation / Sub-step inside a loop or function / Secondary logic)
    ## Generate inspection dataframes to allow structural tracking in interactive notebooks
    go_context_df = pd.DataFrame(go_profile_records)
    
    ## Heading 2 (Specific operation / Sub-step inside a loop or function / Secondary logic)
    ## Execute dataframe masking utilizing normalized mapping target intersections
    filtered_marker_df = working_df[working_df['cellontology_id'].isin(mapped_cl_ids)].copy()
    
    logger.info("Filtering complete. Rows before: %d, Rows remaining: %d", len(working_df), len(filtered_marker_df))
    return filtered_marker_df, go_context_df
```

```python
# Uruchomienie funkcji filtrującej
db_df = pd.read_excel('resources/databases/Cell_marker_All.xlsx')
filtered_markers_df, go_context_df = filter_markers_by_go_validity(db_df, 'resources/databases/go-basic.obo')

# 1. Zobacz ile masz terminów w poszczególnych klasach (procesy, komponenty, funkcje molekularne)
print(go_context_df['go_namespace'].value_counts())

# 2. Wyświetl opisy mechanistyczne, żeby zweryfikować czy zawierają pożądane informacje
go_context_df[['go_term_name', 'go_namespace', 'go_definition']].head(15)
```
